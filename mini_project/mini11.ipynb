```python
import matplotlib.pyplot as plt
import scipy.cluster.hierarchy as sch
```

```python
plt.figure(figsize=(12, 7))
linkage_matrix = sch.linkage(scaled_features, method='ward')
sch.dendrogram(linkage_matrix)
plt.title('Dendrogram')
plt.xlabel('Data Points')
plt.ylabel('Euclidean Distances')
plt.show()
```

```python
from sklearn.cluster import AgglomerativeClustering
```

```python
hc = AgglomerativeClustering(n_clusters=3, metric='euclidean', linkage='ward')
hc_clusters = hc.fit_predict(scaled_features)
```

```python
df['HC_Cluster'] = hc_clusters
```

```python
from sklearn.metrics import silhouette_score
hc_silhouette = silhouette_score(scaled_features, hc_clusters)
print(hc_silhouette)
```

```python
import seaborn as sns
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df, x='petal_length', y='petal_width', hue='HC_Cluster', palette='viridis')
plt.title('Hierarchical Clustering Results')
plt.show()
```
